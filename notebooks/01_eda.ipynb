# Analyse Exploratoire des Données (EDA) - Amazon Sentiment

Ce notebook présente une analyse des données de sentiment générées pour le projet DeepLearn.

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import os
```

## Chargement des données

```python
df = pd.read_csv(os.path.join('..', 'data', 'amazon_reviews.csv'))
df.head()
```

## Visualisation de la distribution des notes (Stars)

```python
fig_stars = px.histogram(df, x='stars', nbins=5, title='Distribution des Notes (Stars)',
                         labels={'stars': 'Etoiles', 'count': 'Nombre de reviews'},
                         color='stars', color_discrete_sequence=px.colors.qualitative.Pastel)
fig_stars.show()
```

## Répartition des Sentiments

```python
sentiment_counts = df['sentiment'].value_counts().reset_index()
sentiment_counts.columns = ['sentiment', 'count']
fig_sentiment = px.pie(sentiment_counts, values='count', names='sentiment', 
                       title='Répartition des Sentiments',
                       color_discrete_sequence=px.colors.qualitative.Safe)
fig_sentiment.show()
```

## Analyse par Produit

```python
product_sentiment = df.groupby(['product_name', 'sentiment']).size().reset_index(name='count')
fig_product = px.bar(product_sentiment, x='product_name', y='count', color='sentiment',
                     title='Sentiments par Produit', barmode='group',
                     labels={'product_name': 'Produit', 'count': 'Nombre de reviews'})
fig_product.show()
```
